```python
import pandas as pd 
import numpy as np
```

```python
customers = pd.read_csv(r"../01_Raw_Data/crm_50000_customers_dirty_v3.csv")
orders = pd.read_csv(r"../01_Raw_Data/orders_300k_dirty.csv")
support = pd.read_csv(r"../01_Raw_Data/support_tickets_30000_dirty.csv")
products = pd.read_csv(r"../01_Raw_Data/product_catalog_dirty_30pct.csv")
clickstream = pd.read_csv(r"../01_Raw_Data/clickstream_500k_events.csv")
```

```python
print("Customers:", customers.shape)
print("Orders:", orders.shape)
print("Support Tickets:", support.shape)
print("Products:", products.shape)
print("Clickstream:", clickstream.shape)
```

```python
print("===CUSTOMERS===")
print(customers.columns.tolist())
print("\n===ORDERS===")
print(orders.columns.tolist())
print("\n===SUPPORT TICKETS===")
print(support.columns.tolist())
print("\n===PRODUCTS===")
print(products.columns.tolist())
print("\n===CLICKSTREAM===")
print(clickstream.columns.tolist())
```

```python
datasets = {"Customers" : customers,
            "Orders" : orders,
            "Suppor Tickets" : support,
            "Products" : products,
            "Clickstream" : clickstream}

for name, df in datasets.items():
    print(f"\n==={name}===")
    missing = df.isnull().sum()
    missing_pct = (missing/len(df)*100).round(2)

    audit = pd.DataFrame({"Missing Count" : missing, 
              "Missing %" : missing_pct})

    print(audit[audit["Missing Count"] > 0])
```

```python
for name, df in datasets.items():
    duplicate_count = df.duplicated().sum()
    duplicate_pct = (duplicate_count/len(df)*100).round(2)

    print(f"{name :}")
    print(f"Duplicate Rows : {duplicate_count}")
    print(f"Duplicate % : {duplicate_pct}%")
    print()
```

```python
key_columns = {
    "Customers": ("customer_id", customers),
    "Orders": ("order_id", orders),
    "Support Tickets": ("ticket_id", support),
    "Products": ("product_id", products),
    "Clickstream": ("event_id", clickstream)
}

for name, (key, df) in key_columns.items():
    duplicate_ids = df[key].duplicated().sum()
    unique_ids = df[key].nunique()
    
    print(f"===== {name} =====")
    print(f"Key: {key}")
    print(f"Total rows: {len(df)}")
    print(f"Unique IDs: {unique_ids}")
    print(f"Duplicate IDs: {duplicate_ids}")
    print()
```

```python
print("===== ORDERS → CUSTOMERS =====")
print("Orders with invalid customer_id:",
      (~orders["customer_id"].isin(customers["customer_id"])).sum())

print("\n===== ORDERS → PRODUCTS =====")
print("Orders with invalid product_id:",
      (~orders["product_id"].isin(products["product_id"])).sum())

print("\n===== SUPPORT → CUSTOMERS =====")
print("Support tickets with invalid customer_id:",
      (~support["customer_id"].isin(customers["customer_id"])).sum())

print("\n===== CLICKSTREAM → CUSTOMERS =====")
print("Clickstream with invalid customer_id:",
      (~clickstream["customer_id"].isin(customers["customer_id"])).sum())
```

```python
print("===== ORDERS =====")
print("Negative order amounts:", (pd.to_numeric(
    orders["order_amount"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
) < 0).sum())

print("Zero order amounts:", (pd.to_numeric(
    orders["order_amount"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
) == 0).sum())

print("Negative quantities:", (pd.to_numeric(
    orders["quantity"].astype(str),
    errors="coerce"
) < 0).sum())

print("Zero quantities:", (pd.to_numeric(
    orders["quantity"].astype(str),
    errors="coerce"
) == 0).sum())


print("\n===== PRODUCTS =====")
print("Negative prices:", (pd.to_numeric(
    products["price"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
) < 0).sum())

print("Zero prices:", (pd.to_numeric(
    products["price"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
) == 0).sum())


print("\n===== SUPPORT =====")
print("Negative resolution time:",
      (support["resolution_time_hours"] < 0).sum())
```
