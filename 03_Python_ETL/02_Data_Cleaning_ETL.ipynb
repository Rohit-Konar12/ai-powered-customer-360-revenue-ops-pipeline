```python
orders_clean = orders.copy()
print(orders_clean.shape)
print(orders_clean.head())
```

```python
orders_clean = orders_clean.drop_duplicates()
print("Rows after removing exact duplicates : ", len(orders_clean))
```

```python
orders_clean["order_amount"] = (orders_clean["order_amount"].astype(str).str.replace(",", "", regex=False))
orders_clean["order_amount"] = pd.to_numeric(orders_clean["order_amount"], errors="coerce")
print(orders_clean["order_amount"].dtype)
print(orders_clean["order_amount"].head(10))
```

```python
orders_clean["order_date"] = pd.to_datetime(orders_clean["order_date"], errors="coerce")
print(orders_clean["order_date"].dtype)
print(orders_clean["order_date"].head(10))
```

```python
orders_clean["payment_method"] = (orders_clean["payment_method"].str.strip().str.lower())
print(orders_clean["payment_method"].value_counts(dropna=False))
```

```python
payment_map = {"crad" : "card", "c@rd" : "card", "wall-et" : "wallet", "cd" : "card"}
orders_clean["payment_method"] = (orders_clean["payment_method"].replace(payment_map))
print(orders_clean["payment_method"].value_counts(dropna=False))
```

```python
orders_clean["status"] = (orders_clean["status"].str.strip().str.lower())
print(orders_clean["status"].value_counts(dropna=False))
```

```python
status_map = {"ref" : "refunded", "suc" : "success", "fail" : "failed"}
orders_clean["status"] = (orders_clean["status"].replace(status_map))
print(orders_clean["status"].value_counts(dropna=False))
```

```python
orders_clean["quantity"] = pd.to_numeric(orders_clean["quantity"], errors="coerce")
orders_clean["quantity"] = orders_clean["quantity"].astype("Int64")
print(orders_clean["quantity"].dtype)
print(orders_clean["quantity"].value_counts(dropna=False).sort_index())
```

```python
orders_clean.loc[orders_clean["quantity"] <= 0, "quantity"] = pd.NA
print(orders_clean["quantity"].value_counts(dropna=False).sort_index())
```

```python
orders_clean.loc[orders_clean["order_amount"] <=0, "order_amount"] = np.nan
print(orders_clean["order_amount"].isna().sum())
print(orders_clean["order_amount"].describe())
```

```python
print(orders_clean.dtypes)
```

```python
print(orders_clean.isnull().sum())
```
