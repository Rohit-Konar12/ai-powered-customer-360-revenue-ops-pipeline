```python
import pandas as pd
import numpy as np
customers = pd.read_csv(r"D:\Customer_360\Cleaned_CSV\customers_clean.csv")
orders = pd.read_csv(r"D:\Customer_360\Cleaned_CSV\orders_clean.csv")
support = pd.read_csv(r"D:\Customer_360\Cleaned_CSV\support_clean.csv")
products = pd.read_csv(r"D:\Customer_360\Cleaned_CSV\products_clean.csv")
clickstream = pd.read_csv(r"D:\Customer_360\Cleaned_CSV\clickstream_clean.csv")
print("All cleaned datasets are loaded successfully.")
```

```python
print("CUSTOMERS")
print(customers.shape)
print(customers.columns.tolist())

print("\nORDERS")
print(orders.shape)
print(orders.columns.tolist())

print("\nSUPPORT")
print(support.shape)
print(support.columns.tolist())

print("\nPRODUCTS")
print(products.shape)
print(products.columns.tolist())

print("\nCLICKSTREAM")
print(clickstream.shape)
print(clickstream.columns.tolist())
```

```python
print("Orders - Neumeric Summary")
print(orders[["order_amount", "quantity"]].describe())

print("\nProducts - Neumeric Summary")
print(products[["price"]].describe())

print("\nSupport - Neumeric Summary")
print(support[["resolution_time_hours"]].describe())
```

```python
print("GENDER")
print(customers["gender"].value_counts())

print("\nSOURCE")
print(customers["source"].value_counts())

print("\nCOUNTRY")
print(customers["country"].value_counts())

print("\nTOP 10 CITIES")
print(customers["city"].value_counts().head(10))
```

```python
total_revenue = orders["order_amount"].sum()
total_orders = orders["order_id"].nunique()
average_order_value = orders["order_amount"].mean()
total_quantity = orders["quantity"].sum()
unique_buying_customers = orders["customer_id"].nunique()

print("=== BUSINESS KPI BASELINE ===")
print("Total Revenue:", total_revenue)
print("Total Orders:", total_orders)
print("Average Order Value:", average_order_value)
print("Total Quantity Sold:", total_quantity)
print("Unique Buying Customers:", unique_buying_customers)
```

```python
print("Order amount distribution.")
print(orders["order_amount"].describe(percentiles=[0.10, 0.25, 0.50, 0.75, 0.90, 0.95, 0.99]))
```

```python
print(orders["order_date"].dtype)
```

```python
orders["order_date"] = pd.to_datetime(orders["order_date"], errors="coerce")
monthly_revenue = (orders.dropna(subset=["order_date", "order_amount"]).assign(month=lambda x: x["order_date"].dt.to_period("M").astype(str)).groupby("month")["order_amount"].sem())

print("Monthly Revenue")
print(monthly_revenue)
```

```python
orders_per_customer = (orders.groupby("customer_id")["order_id"].nunique())
print("Orders per customer")
print(orders_per_customer.describe(percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99]))
```

```python
customer_order_counts = (orders.groupby("customer_id")["order_id"].nunique())
one_time_customers = (customer_order_counts == 1).sum()
repeat_customers = (customer_order_counts > 1).sum()
print("One-time customers: ", one_time_customers)
print("Repeat customers: ", repeat_customers)
print("\nRepeat cutomers rate")
print(round(repeat_customers/len(customer_order_counts)*100, 2), "%")
```

```python
customer_revenue = (orders.groupby("customer_id")["order_amount"].sum())
print("Customers Revenue Distribution")
print(customer_revenue.describe(percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99]))
```

```python
customer_revenue_sorted = customer_revenue.sort_values(ascending=False)
total_customer_revenue = customer_revenue_sorted.sum()
top_10_pct_count = int(len(customer_revenue_sorted)*0.10)
top_10_revenue = customer_revenue_sorted.head(top_10_pct_count).sum()
top_10_revenue_share = (top_10_revenue/total_customer_revenue*100)
print("Top 10% customer revenue share:", round(top_10_revenue_share, 2), "%")
```

```python
category_revenue = (orders.merge(products[["product_id", "category"]],on="product_id",how="left").groupby("category")["order_amount"].sum().sort_values(ascending=False))
print("Revenue by Product Category")
print(category_revenue)
```

```python
category_quantity = (orders.merge(products[["product_id", "category"]],on="product_id",how="left").groupby("category")["quantity"].sum().sort_values(ascending=False))
print("Quantity Sold by Product Category")
print(category_quantity)
```

```python
category_performance = (orders.merge(products[["product_id", "category"]],on="product_id",how="left").groupby("category").agg(revenue=("order_amount", "sum"),quantity=("quantity", "sum")))
category_performance["revenue_per_unit"] = (category_performance["revenue"] /category_performance["quantity"])
print("Category Revenue per Unit")
print(category_performance.sort_values("revenue_per_unit", ascending=False))
```

```python
issue_distribution = (support["issue_type"].value_counts())
print("Support Tickets by Issue Type")
print(issue_distribution)
```

```python
sentiment_distribution = (support["sentiment"].value_counts())
print("Support Tickets by Sentiment")
print(sentiment_distribution)
```

```python
issue_sentiment = pd.crosstab(support["issue_type"],support["sentiment"])
print("Issue Type vs Sentiment")
print(issue_sentiment)
```

```python
event_distribution = (clickstream["event_type"].value_counts())
print("Clickstream Events")
print(event_distribution)
```

```python
customer_engagement = (clickstream.dropna(subset=["customer_id"]).groupby("customer_id")["event_id"].nunique().rename("engagement_events"))
customer_orders = (orders.groupby("customer_id")["order_id"].nunique().rename("order_count"))
engagement_orders = pd.concat([customer_engagement, customer_orders],axis=1).dropna()
print("Customer Engagement vs Orders")
print(engagement_orders.describe())
```

```python
engagement_orders["engagement_level"] = pd.cut(engagement_orders["engagement_events"],bins=[0, 5, 10, float("inf")],labels=["Low", "Medium", "High"])
engagement_comparison = (engagement_orders.groupby("engagement_level", observed=True)["order_count"].agg(["count", "mean", "median"]))
print("Engagement Level vs Order Count")
print(engagement_comparison)
```

```python
frequency_revenue = (orders.groupby("customer_id").agg(order_count=("order_id", "nunique"),revenue=("order_amount", "sum")))
frequency_revenue["customer_segment"] = pd.cut(frequency_revenue["order_count"],bins=[0, 2, 5, 10, float("inf")],labels=["1-2 Orders", "3-5 Orders", "6-10 Orders", "11+ Orders"])
frequency_comparison = (frequency_revenue.groupby("customer_segment", observed=True).agg(customers=("order_count", "count"),avg_orders=("order_count", "mean"),avg_revenue=("revenue", "mean")))
print("Purchase Frequency vs Customer Revenue")
print(frequency_comparison)
```

```python
segment_revenue = (frequency_revenue.groupby("customer_segment", observed=True)["revenue"].agg(["sum", "count"]))
segment_revenue["revenue_share_pct"] = (segment_revenue["sum"]/ segment_revenue["sum"].sum()* 100)
print("Revenue Contribution by Customer Segment")
print(segment_revenue)
```

```python
customer_tickets = (support.groupby("customer_id")["ticket_id"].nunique().rename("ticket_count"))
customer_revenue_support = pd.concat([customer_revenue.rename("customer_revenue"),customer_tickets],axis=1).fillna({"ticket_count": 0})
customer_revenue_support["support_level"] = pd.cut(customer_revenue_support["ticket_count"],bins=[-1, 0, 2, 5, float("inf")],labels=["No Tickets", "1-2 Tickets", "3-5 Tickets", "6+ Tickets"])
support_revenue_comparison = (customer_revenue_support.groupby("support_level", observed=True).agg(customers=("ticket_count", "count"),avg_tickets=("ticket_count", "mean"),avg_revenue=("customer_revenue", "mean")))
print("Support Activity vs Customer Revenue")
print(support_revenue_comparison)
```

```python
sentiment_resolution = (support.groupby("sentiment")["resolution_time_hours"].agg(["count", "mean", "median"]).sort_values("mean", ascending=False))
print("Sentiment vs Resolution Time")
print(sentiment_resolution)
```

```python
issue_resolution = (support.groupby("issue_type")["resolution_time_hours"].agg(["count", "mean", "median"]).sort_values("mean", ascending=False))
print("Issue Type vs Resolution Time")
print(issue_resolution)
```

```python
support_activity_sentiment = (support.groupby(["customer_id", "sentiment"])["ticket_id"].nunique().reset_index())
print("Customer Support Activity by Sentiment")
print(support_activity_sentiment.groupby("sentiment")["ticket_id"].agg(["count", "mean", "median"]))
```

```python
engagement_support = pd.concat([customer_engagement.rename("engagement_events"),customer_tickets.rename("ticket_count")],axis=1).fillna(0)
engagement_support["engagement_level"] = pd.cut(engagement_support["engagement_events"],bins=[-1, 5, 10, float("inf")],labels=["Low", "Medium", "High"])
engagement_support_comparison = (engagement_support.groupby("engagement_level", observed=True).agg(customers=("ticket_count", "count"),avg_engagement=("engagement_events", "mean"),avg_tickets=("ticket_count", "mean")))
print("Engagement Level vs Support Activity")
print(engagement_support_comparison)
```

```python
customer_360 = pd.concat([customer_engagement.rename("engagement_events"),customer_orders.rename("order_count"),customer_tickets.rename("ticket_count")],axis=1).fillna(0)
customer_360["engagement_level"] = pd.cut(customer_360["engagement_events"],bins=[-1, 5, 10, float("inf")],labels=["Low", "Medium", "High"])
customer_360["purchase_level"] = pd.cut(customer_360["order_count"],bins=[0, 2, 5, 10, float("inf")],labels=["1-2 Orders", "3-5 Orders", "6-10 Orders", "11+ Orders"])
customer_360_summary = (customer_360.groupby(["engagement_level", "purchase_level"], observed=True).agg(customers=("order_count", "count"),avg_engagement=("engagement_events", "mean"),avg_orders=("order_count", "mean"),avg_tickets=("ticket_count", "mean")))
print("Customer 360 Segmentation")
print(customer_360_summary)
```
